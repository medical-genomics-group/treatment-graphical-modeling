```python
from pathlib import Path
import pandas as pd

%load_ext autoreload
%autoreload 2

from prep_regenie_inputs import (
    make_normal_inputs,
    make_int_inputs,
    prepare_regenie_inputs_from_manifest
)

bp_pooled = "<BP_POOLED_TSV>"
ldl_pooled = "<LDL_POOLED_TSV>"

bp_split = "<BP_AGE5_TSV>"
ldl_split = "<LDL_AGE5_TSV>"

ukb_covs = "<UKB_COVARS_COV>"
ukb_namecovs = "<UKB_COVARS_NAMECOV>"
fam = "<FAM_FILE>"
```

```python
out_root = "<OUT_ROOT>"
```

```python
out_root = Path(out_root)
(out_root / "regenie_phenos").mkdir(parents=True, exist_ok=True)
(out_root / "regenie_covars").mkdir(parents=True, exist_ok=True)

classic = make_normal_inputs(out_root, bp_pooled, ldl_pooled, bp_split, ldl_split)
interx  = make_int_inputs(out_root, bp_pooled, ldl_pooled)
```

```python
classic.to_csv("regenie-runs.tsv")
```

```python
interx.to_csv("regenie-runs-int.tsv")
```

```python
prepare_regenie_inputs_from_manifest(
    manifest_tsv="regenie-runs.tsv",
    out_root=out_root,
    ukb_covs=ukb_covs,
    ukb_namecovs=ukb_namecovs,
    fam=fam,
)
```

```python
prepare_regenie_inputs_from_manifest(
    manifest_tsv="regenie-runs-int.tsv",
    out_root=out_root,
    ukb_covs=ukb_covs,
    ukb_namecovs=ukb_namecovs,
    fam=fam,
)
```

```python
pairs_all = pd.concat([
    classic.assign(interaction_var=".")[
        ["run_id","scope","trait","phenotype","covars_in_model","pheno_path","covar_path","interaction_var"]
    ],
    interx[
        ["run_id","scope","trait","phenotype","covars_in_model","pheno_path","covar_path","interaction_var"]
    ]
], ignore_index=True)

pairs_all.to_csv("pairs_all.tsv", sep="\t", index=False)
```
